```python
from sklearn.tree  import DecisionTreeClassifier,plot_tree
from sklearn.preprocessing import OrdinalEncoder ,StandardScaler
from sklearn.model_selection  import train_test_split 
from sklearn.datasets  import make_classification
import pandas as pd 
import numpy as np
```

```python
X,y =  make_classification(n_samples= 100 , n_classes= 2,n_informative=3,class_sep= 1.5,n_features=3,n_redundant=0,n_repeated=0)
X
```

```python
df =  pd.DataFrame(X,columns= ['attendance','score','cgpa'])
```

```python
df['target'] =  y
X =  df.drop(columns='target')
y =  df['target']
```

```python
scaler =  StandardScaler()
```

```python
x_train,x_test,y_train,y_test =  train_test_split(
    X,y,test_size= 0.2
)
```

```python

x_train=  scaler.fit_transform(x_train)
x_test=  scaler.transform(x_test)
```

```python
model =  DecisionTreeClassifier(
    criterion='entropy',
    max_depth=6,
    random_state= 42
)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred =  model.predict(x_test)
```

```python
from sklearn.metrics  import accuracy_score
print("accuarcy : ",accuracy_score(y_test,y_pred))
```

```python
import matplotlib.pyplot as plt 

plt.figure(figsize=(7,8))
plot_tree(model,feature_names=X.columns,filled= True,class_names=['Fail','Pass'])
plt.show()
```
